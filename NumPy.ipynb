## Part 1: Introduction to NumPy

```python
import numpy as np
import time
n = 1_000_000
python_list = list(range(n))
start = time.time()
python_result = [x + 1 for x in python_list]
list_time = time.time() - start
numpy_array = np.arange(n)
start = time.time()
numpy_result = numpy_array + 1
numpy_time = time.time() - start
print("Python List time:", list_time)
print("NumPy time:", numpy_time)
```

NumPy is faster because it performs operations on the entire array at once using compiled code instead of python processing each number one by one in a loop . so it reduces overhead of python loops and numerical operations much more efficient.

Part 15: Text Analysis Using NumPy (NLP Track)

Step 1: From Text to Tokens

```python
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"
tokens2 = sentence2.split()
print(tokens2)
```

here we have used split() to divide the scentence into individual words stores the result in token2.

```python
vocabulary = set(tokens2)
unique_words = len(vocabulary)
print("Vocabulary:", vocabulary)
print("Number of unique words:", unique_words)
```

here we have converted tokens2 into set then removed duplicates then uses len() to count unique words.

Step 2: Building a Vocabulary

```python
vocabulary2 = list(set(tokens2))
bow_vector = [tokens2.count(word) for word in vocabulary2]
print("Vocabulary:", vocabulary2)
print("Bag-of-Words vector:", bow_vector)
```

Prediction No word will have a count of 2 cause every word in sentence appears only once.

Step 3:
Word Frequency Vector (Bag-of-Words)


```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
sentence1 = "I love learning Python programming"
sentence2 = "I enjoy studying Python programming"
sentence3 = "The cat is sleeping on the sofa"
sentence4 = "The weather is sunny today"
vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform([sentence1, sentence2, sentence3, sentence4])
similar_pair = cosine_similarity(vectors[0], vectors[1])[0][0]
different_pair = cosine_similarity(vectors[2], vectors[3])[0][0]
print("Similarity of similar sentences:", similar_pair)
print("Similarity of different sentences:", different_pair)
```

Cosine similarity tells the direction of two vectors. Similar sentences contain more common words, so their vectors point in similar directions and the value is closer to 1. Unrelated sentences have fewer common words, so their vectors are more different and the value is closer to 0.

Step 4: Comparing Two Sentences (Cosine Similarity)

```python
import numpy as np
paragraph = """
Python is a powerful programming language.
Python is easy to learn and Python is widely used.
Programming with Python helps students learn data science.
"""
tokens = paragraph.lower().split()
vocabulary = list(set(tokens))
bag_of_words_vector = np.array([tokens.count(word) for word in vocabulary])
sorted_indices = np.argsort(bag_of_words_vector)[::-1]
top_3_indices = sorted_indices[:3]
print("Top 3 most frequent words:")
for i in top_3_indices:
    print(vocabulary[i], ":", bag_of_words_vector[i])
```

1st we do tokenaise the para then turn into vocab then counts how often each word appears . np.argsort() is udes to sort word by indices by frequency and display top 3 most frequent words.

Step 5: Most Frequent Words

```python
import numpy as np
sentence1 = "I love learning Python programming"
sentence2 = "I enjoy learning Python programming"
sentence3 = "The cat is sleeping on the sofa"
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()
vocabulary = list(set(tokens1 + tokens2 + tokens3))
vector1 = np.array([tokens1.count(word) for word in vocabulary])
vector2 = np.array([tokens2.count(word) for word in vocabulary])
vector3 = np.array([tokens3.count(word) for word in vocabulary])
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
similarity_12 = cosine_similarity(vector1, vector2)
similarity_13 = cosine_similarity(vector1, vector3)
similarity_23 = cosine_similarity(vector2, vector3)
print("Similarity (Sentence 1, Sentence 2):", similarity_12)
print("Similarity (Sentence 1, Sentence 3):", similarity_13)
print("Similarity (Sentence 2, Sentence 3):", similarity_23)
scores = {
    "Sentence 1 & Sentence 2": similarity_12,          #AP24110010586
    "Sentence 1 & Sentence 3": similarity_13,
    "Sentence 2 & Sentence 3": similarity_23
}
most_similar_pair = max(scores, key=scores.get)
print("\nMost similar pair:", most_similar_pair)
print("Similarity score:", scores[most_similar_pair])
```
